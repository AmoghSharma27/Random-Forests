```python
from RandomForest import RandomForests
import math
import pandas as pd
import BinaryTree
```

```python
def import_csv(file_name="./database/All_Pokemon.csv"):
    return pd.read_csv(file_name)


def main():
    data = import_csv()

    # 80-20 split in testing and training data
    num_rows = len(data)
    train_data = data.head(math.ceil(num_rows * 0.8))
    test_data = data.tail(math.floor(num_rows * 0.2))

    random_forests = RandomForests(train_data, 100)

    # Calculate how correct
    correct = 0
    amount = 0
    for i in range(len(test_data)):
        pred = random_forests.predict(test_data.iloc[i])
        if pred == test_data.iloc[i]["Type 1"]:
            correct += 1
        else:
            print("Pokemon:\t", test_data.iloc[i]["Name"])
            print("Actual:\t", test_data.iloc[i]["Type 1"])
            print("Predicted:\t", pred)
            print()
        amount += 1
    print("Amount:\t", amount)
    print(f"Accuracy:\t{(correct/amount) * 100.00}%")
```

```python
if __name__ == "__main__":
    main()
```
